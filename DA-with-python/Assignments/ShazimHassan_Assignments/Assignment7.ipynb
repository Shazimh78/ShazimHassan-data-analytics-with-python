5) Use loc and iloc to (a) get the 2nd row, (b) get Name of the 4th employee, (c) get rows 1 to 3 with columns Name and City.

6) Add a new column 'Bonus' that is 10% of Salary and a column 'Total' = Salary + Bonus. Then rename the column 'City' to 'Location' and drop the 'Bonus' column.

```python
import pandas as pd

data = {
    "Name": ["Ali", "Sara", "Ahmed", "Zara", "Bilal", "Hina"],
    "Age": [25, 32, 28, 35, 22, 31],
    "City": ["Delhi", "Lahore", "Delhi", "Karachi", "Delhi", "Delhi"],
    "Salary": [55000, 72000, 38000, 90000, 40000, 65000]
}

df = pd.DataFrame(data)

df["Bonus"] = df["Salary"] * 0.10
df["Total"] = df["Salary"] + df["Bonus"]
print("After adding Bonus and Total:")
print(df)

df = df.rename(columns={"City": "Location"})
print("After renaming City to Location:")
print(df)

df = df.drop("Bonus", axis=1)
print("After dropping Bonus:")
print(df)
```

7) Create a DataFrame having some missing values (NaN). Find the count of missing values in each column, fill numeric NaNs with the column mean, and drop rows that still have missing values.

```python
import pandas as pd
import numpy as np

data = {
    "Name": ["Ali", "Sara", "Ahmed", "Zara", "Bilal", "Hina"],
    "Age": [25, np.nan, 28, 35, np.nan, 31],
    "City": ["Delhi", "Lahore", None, "Karachi", "Delhi", "Delhi"],
    "Salary": [55000, 72000, np.nan, 90000, 40000, np.nan]
}

df = pd.DataFrame(data)

print("Original DataFrame:")
print(df)

print("Missing values in each column:")
print(df.isnull().sum())

numeric_cols = df.select_dtypes(include="number").columns
df[numeric_cols] = df[numeric_cols].fillna(df[numeric_cols].mean())

print("After filling numeric NaNs with column mean:")
print(df)

df = df.dropna()

print("After dropping rows with missing values:")
print(df)
```

8) Sort the DataFrame by Salary in descending order, and then by Age in ascending order. Also print the top 3 highest-paid employees using nlargest().

```python
import pandas as pd

data = {
    "Name": ["Ali", "Sara", "Ahmed", "Zara", "Bilal", "Hina"],
    "Age": [25, 32, 28, 35, 22, 31],
    "City": ["Delhi", "Lahore", "Delhi", "Karachi", "Delhi", "Delhi"],
    "Salary": [55000, 72000, 38000, 90000, 40000, 65000]
}

df = pd.DataFrame(data)

print("Sorted by Salary (descending), then Age (ascending):")
print(df.sort_values(by=["Salary", "Age"], ascending=[False, True]))

print("Sorted by Age (ascending):")
print(df.sort_values(by="Age", ascending=True))

print("Top 3 highest-paid employees:")
print(df.nlargest(3, "Salary"))
```

9) Using groupby, find the average salary per City, the number of employees per City, and the maximum salary per City. Show the result using agg() with multiple functions.

```python
import pandas as pd

data = {
    "Name": ["Ali", "Sara", "Ahmed", "Zara", "Bilal", "Hina"],
    "Age": [25, 32, 28, 35, 22, 31],
    "City": ["Delhi", "Lahore", "Delhi", "Karachi", "Delhi", "Delhi"],
    "Salary": [55000, 72000, 38000, 90000, 40000, 65000]
}

df = pd.DataFrame(data)

print("Average salary per City:")
print(df.groupby("City")["Salary"].mean())

print("Number of employees per City:")
print(df.groupby("City")["Name"].count())

print("Maximum salary per City:")
print(df.groupby("City")["Salary"].max())

print("Combined using agg():")
print(df.groupby("City")["Salary"].agg(["mean", "count", "max"]))
```

10) Create two DataFrames (employees and departments) with a common column 'DeptID'. Merge them using inner and left joins, then concatenate two DataFrames of the same structure using pd.concat() and remove duplicate rows.

```python
import pandas as pd

employees = pd.DataFrame({
    "Name": ["Ali", "Sara", "Ahmed", "Zara", "Bilal"],
    "Salary": [55000, 72000, 38000, 90000, 40000],
    "DeptID": [1, 2, 1, 3, 4]
})

departments = pd.DataFrame({
    "DeptID": [1, 2, 3],
    "DeptName": ["Sales", "IT", "HR"]
})

inner_join = pd.merge(employees, departments, on="DeptID", how="inner")
print("Inner join:")
print(inner_join)

left_join = pd.merge(employees, departments, on="DeptID", how="left")
print("Left join:")
print(left_join)

df1 = pd.DataFrame({
    "Name": ["Ali", "Sara", "Ahmed"],
    "Salary": [55000, 72000, 38000],
    "DeptID": [1, 2, 1]
})

df2 = pd.DataFrame({
    "Name": ["Sara", "Zara", "Bilal"],
    "Salary": [72000, 90000, 40000],
    "DeptID": [2, 3, 4]
})

combined = pd.concat([df1, df2])
print("After concat:")
print(combined)

combined = combined.drop_duplicates().reset_index(drop=True)
print("After removing duplicates:")
print(combined)
```
