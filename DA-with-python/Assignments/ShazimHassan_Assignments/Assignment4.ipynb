person = {
"name": "Ali",
"age": 25,
"city": "Karachi"
}

Print keys and values using loop.

```python
person = {
    "name": "Ali",
    "age": 25,
    "city": "Karachi"
}

for key, value in person.items():
    print(key, ":", value)
```

Q4 — Count Frequency of Numbers

numbers = [1,2,2,3,3,3,4,4,4,4]

Create frequency dictionary.

```python
numbers = [1, 2, 2, 3, 3, 3, 4, 4, 4, 4]

frequency = {}
for num in numbers:
    if num in frequency:
        frequency[num] += 1
    else:
        frequency[num] = 1

print(frequency)
```

Q5 — Word Counter

Take sentence input and count word frequency.

```python
sentence = input("Enter a sentence: ")
words = sentence.split()

word_count = {}
for word in words:
    if word in word_count:
        word_count[word] += 1
    else:
        word_count[word] = 1

print(word_count)
```

Q6 — Student Marks & Average

marks = {"Math":80, "English":70, "Physics":90}

Calculate total and average.

```python
marks = {"Math": 80, "English": 70, "Physics": 90}

total = sum(marks.values())
average = total / len(marks)

print("Total:", total)
print("Average:", average)
```

Q7 — Highest Scoring Student

students = {
"Ali":85,
"Sara":92,
"Ahmed":78,
"Zara":95
}

Find highest scoring student.

```python
students = {
    "Ali": 85,
    "Sara": 92,
    "Ahmed": 78,
    "Zara": 95
}

highest_name = max(students, key=students.get)
highest_score = students[highest_name]

print("Highest scoring student:", highest_name, "with", highest_score, "marks")
```

Q8 — Merge Dictionaries

dict1 = {"a":1,"b":2}
dict2 = {"c":3,"d":4}

Merge and print result.

```python
dict1 = {"a": 1, "b": 2}
dict2 = {"c": 3, "d": 4}

merged_dict = {**dict1, **dict2}

print(merged_dict)
```

Q9 — Mini Project: Contact Book

Add 3 contacts, search by name, print phone or Contact not found.

```python
contacts = {}

for i in range(3):
    name = input("Enter contact name: ")
    phone = input("Enter phone number: ")
    contacts[name] = phone

search_name = input("Enter name to search: ")

if search_name in contacts:
    print("Phone:", contacts[search_name])
else:
    print("Contact not found")
```
