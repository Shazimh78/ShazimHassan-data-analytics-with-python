3. Write a program in which use all the operators we can use in Python

```python
a = 10
b = 3

print("Arithmetic Operators")
print("Addition:", a + b)
print("Subtraction:", a - b)
print("Multiplication:", a * b)
print("Division:", a / b)
print("Floor Division:", a // b)
print("Modulus:", a % b)
print("Exponentiation:", a ** b)

print("\nComparison Operators")
print("Equal to:", a == b)
print("Not equal to:", a != b)
print("Greater than:", a > b)
print("Less than:", a < b)
print("Greater than or equal to:", a >= b)
print("Less than or equal to:", a <= b)

print("\nLogical Operators")
x = True
y = False
print("AND:", x and y)
print("OR:", x or y)
print("NOT:", not x)

print("\nAssignment Operators")
c = 5
c += 2 
print("c after +=:", c)
c -= 1
print("c after -=:", c)
c *= 3
print("c after *=:", c)
c //= 2
print("c after //=:", c)

print("\nBitwise Operators")
print("AND:", a & b)
print("OR:", a | b)
print("XOR:", a ^ b)
print("NOT:", ~a)
print("Left Shift:", a << 1)
print("Right Shift:", a >> 1)

print("\nIdentity Operators")
p = [1, 2, 3]
q = [1, 2, 3]
r = p
print("p is q:", p is q)     
print("p is r:", p is r)    
print("p is not q:", p is not q)
```

4. Completes the following steps of small task:
    - Mention Marks of English , Islamiat and Maths out of 100 in 3 different variables
    - Mention Variable of Total Marks and assign 300 to it
    - Calculate Percentage

```python
english = 85
islamiat = 90
maths = 78

total_marks = 300

obtained_marks = english + islamiat + maths
percentage = (obtained_marks / total_marks) * 100

print("Marks in English:", english)
print("Marks in Islamiat:", islamiat)
print("Marks in Maths:", maths)
print("Total Marks:", total_marks)
print("Obtained Marks:", obtained_marks)
print("Percentage:", percentage,"%")
```

5. Take the values of two variables `a` and `b`, then swap their values without using a third variable and print the result before and after swapping
