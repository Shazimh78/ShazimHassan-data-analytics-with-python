1. Write a program that accepts a list from user and print the alternate element of list.

```python
my_list = []
n = int(input("Enter number of elements: "))

for i in range(n):
    element = input("Enter element: ")
    my_list.append(element)

print("Alternate elements:")
for i in range(0, len(my_list), 2):
    print(my_list[i])
```

2. Write a program that accepts a list from user. Your program should reverse the content of list and
display it. Do not use reverse() method. 

```python
my_list = []
n = int(input("Enter number of elements: "))

for i in range(n):
    element = input("Enter element: ")
    my_list.append(element)

reversed_list = my_list[::-1]

print("Reversed list:")
print(reversed_list)
```

 3. Find and display the largest number of a list without using built-in function max(). Your program
should ask the user to input values in list from keyboard. 

```python
my_list = []
n = int(input("Enter number of elements: "))

for i in range(n):
    element = int(input("Enter element: "))
    my_list.append(element)

largest = my_list[0]
for num in my_list:
    if num > largest:
        largest = num

print("Largest number:", largest)
```

 4. Write a program that rotates the element of a list so that the element at the first index moves to the
second index, the element in the second index moves to the third index, etc., and the element in the last
index moves to the first index. 

```python
my_list = []
n = int(input("Enter number of elements: "))

for i in range(n):
    element = input("Enter element: ")
    my_list.append(element)

rotated_list = [my_list[-1]] + my_list[:-1]

print("Rotated list:")
print(rotated_list)
```

5. Write a program that input a string and ask user to delete a given word from a string. 

```python
text = input("Enter a string: ")
word_to_delete = input("Enter word to delete: ")

words = text.split()
words = [word for word in words if word != word_to_delete]
result = " ".join(words)

print("Updated string:", result)
```

6. Write a program that reads a string from the user containing a date in the form mm/dd/yyyy. It 
should print the date in the form March 12, 2021. 

```python
date_string = input("Enter date (mm/dd/yyyy): ")
month, day, year = date_string.split("/")

months = ["January", "February", "March", "April", "May", "June", "July", "August", "September", "October", "November", "December"]

month_name = months[int(month) - 1]

print(f"{month_name} {int(day)}, {year}")
```

7. Write a program with a function that accepts a string from keyboard and create a new string after 
converting character of each word capitalized. For instance, if the sentence is "stop and smell the roses." 
the output should be "Stop And Smell The Roses"

```python
def capitalize_words(sentence):
    words = sentence.split()
    capitalized_words = [word.capitalize() for word in words]
    return " ".join(capitalized_words)

sentence = input("Enter a sentence: ")
result = capitalize_words(sentence)
print(result)
```

8. Find the sum of each row of matrix of size m x n. For example for the following matrix output will be 
like this : 

```python
matrix = []
rows = int(input("Enter number of rows: "))
cols = int(input("Enter number of columns: "))

for i in range(rows):
    row = []
    for j in range(cols):
        value = int(input(f"Enter element [{i}][{j}]: "))
        row.append(value)
    matrix.append(row)

for i in range(rows):
    row_sum = sum(matrix[i])
    print(f"Sum of row {i+1} = {row_sum}")
```

9. Write a program to add two matrices of size n x m.

```python
rows = int(input("Enter number of rows: "))
cols = int(input("Enter number of columns: "))

print("Enter first matrix:")
matrix1 = []
for i in range(rows):
    row = []
    for j in range(cols):
        value = int(input(f"Enter element [{i}][{j}]: "))
        row.append(value)
    matrix1.append(row)

print("Enter second matrix:")
matrix2 = []
for i in range(rows):
    row = []
    for j in range(cols):
        value = int(input(f"Enter element [{i}][{j}]: "))
        row.append(value)
    matrix2.append(row)

result = []
for i in range(rows):
    row = []
    for j in range(cols):
        row.append(matrix1[i][j] + matrix2[i][j])
    result.append(row)

print("Sum of matrices:")
for row in result:
    print(row)
```

10. Write a program to multiply two matrices.

```python
rows1 = int(input("Enter rows of first matrix: "))
cols1 = int(input("Enter columns of first matrix: "))

print("Enter first matrix:")
matrix1 = []
for i in range(rows1):
    row = []
    for j in range(cols1):
        value = int(input(f"Enter element [{i}][{j}]: "))
        row.append(value)
    matrix1.append(row)

rows2 = int(input("Enter rows of second matrix: "))
cols2 = int(input("Enter columns of second matrix: "))

print("Enter second matrix:")
matrix2 = []
for i in range(rows2):
    row = []
    for j in range(cols2):
        value = int(input(f"Enter element [{i}][{j}]: "))
        row.append(value)
    matrix2.append(row)

result = [[0 for _ in range(cols2)] for _ in range(rows1)]

for i in range(rows1):
    for j in range(cols2):
        for k in range(cols1):
            result[i][j] += matrix1[i][k] * matrix2[k][j]

print("Product of matrices:")
for row in result:
    print(row)
```
